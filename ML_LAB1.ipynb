**TASK 2**

```python


print("City Data Missing Values Before Cleaning")
print(city_data.isnull().sum())
if 'AQI_Bucket' in city_data.columns:
    city_data = city_data.drop(columns=['AQI_Bucket'])
numerical_cols_city = ['PM2.5', 'PM10', 'NO', 'NO2', 'NOx', 'NH3', 'CO', 'SO2', 'O3', 'Benzene', 'Toluene', 'Xylene', 'AQI']
for col in numerical_cols_city:
    if city_data[col].isnull().any():
        median_val = city_data[col].median()
        city_data[col].fillna(median_val, inplace=True)

print("City Data  Missing Values After Cleaning ")
print(city_data.isnull().sum())


```

```python
print("Crop Data Missing Values Before Cleaning")
print(crop_data.isnull().sum())

if crop_data['Production'].isnull().any():
    median_production = crop_data['Production'].median()
    crop_data['Production'].fillna(median_production, inplace=True)

print("Crop Data  Missing Values After Cleaning")
print(crop_data.isnull().sum())
```

Missing values in the city air quality dataset were imputed using the column medians

The median was  chosen over the mean because  high value outliers mean is affected by extreme values therefore median was taken for imputation

**TASK 5**

```python

Q1 = city_data['AQI'].quantile(0.25)
Q3 = city_data['AQI'].quantile(0.75)
IQR = Q3 - Q1


upper_fence = Q3 + 1.5 * IQR

print(f"Q1 (25th percentile): {Q1:.2f}")
print(f"Q3 (75th percentile): {Q3:.2f}")
print(f"IQR: {IQR:.2f}")
print(f"Upper Fence (outlier boundary): {upper_fence:.2f}")


outliers_count = city_data[city_data['AQI'] > upper_fence].shape[0]
print(f"\nNumber of implausibly high AQI readings (outliers) detected: {outliers_count}")


city_data_treated = city_data.copy()
city_data_treated['AQI'] = city_data_treated['AQI'].apply(lambda x: upper_fence if x > upper_fence else x)

print(f"\nAQI values after capping will not exceed: {city_data_treated['AQI'].max():.2f}")
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(16, 6))

plt.subplot(1, 2, 1)
sns.histplot(city_data['AQI'], bins=50, kde=True, color='skyblue')
plt.title('Original AQI Distribution (Before Capping)')
plt.xlabel('AQI Value')
plt.ylabel('Frequency')
plt.axvline(x=upper_fence, color='red', linestyle='--', label=f'Upper Fence ({upper_fence:.2f})')
plt.legend()
plt.grid(axis='y', linestyle='--', alpha=0.7)

plt.subplot(1, 2, 2)
sns.histplot(city_data_treated['AQI'], bins=50, kde=True, color='lightcoral')
plt.title('AQI Distribution After Outlier Capping')
plt.xlabel('AQI Value (Capped)')
plt.ylabel('Frequency')
plt.axvline(x=upper_fence, color='red', linestyle='--', label=f'Capped at ({upper_fence:.2f})')
plt.legend()
plt.grid(axis='y', linestyle='--', alpha=0.7)

plt.tight_layout()
plt.show()
```

The Original AQI Distribution plot clearly shows a long tail of high AQI values extending the upper fence indicating the presence of outliers

After Outlier Capping plot this long tail is absent and all values above the calculated upper fence

**TASK 8**

```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

city_to_state_mapping = {
    'ahmedabad': 'gujarat',
    'bengaluru': 'karnataka',
    'chennai': 'tamil nadu',
    'delhi': 'delhi',
    'hyderabad': 'telangana',
    'kolkata': 'west bengal',
    'mumbai': 'maharashtra',
    'jaipur': 'rajasthan',
    'kanpur': 'uttar pradesh',
    'lucknow': 'uttar pradesh',
    'patna': 'bihar',
    'bhopal': 'madhya pradesh',
    'chandigarh': 'chandigarh',
    'amritsar': 'punjab',
    'coimbatore': 'tamil nadu',
    'gurugram': 'haryana',
    'indore': 'madhya pradesh',
    'kochi': 'kerala',
    'shillong': 'meghalaya',
    'visakhapatnam': 'andhra pradesh',
    'guwahati': 'assam',
    'eranakulam': 'kerala',
    'amaravati': 'andhra pradesh',
    'brahmapur': 'odisha',
    'aisbagh': 'uttar pradesh'
}

city_data_treated['State_Name'] = city_data_treated['City'].map(city_to_state_mapping)
city_data_agg = city_data_treated.dropna(subset=['State_Name']).copy()
city_data_agg['State_Name'] = city_data_agg['State_Name'].str.lower().str.strip()
city_data_agg['Year'] = city_data_agg['Date'].dt.year

pollutant_cols = ['PM2.5', 'PM10', 'NO', 'NO2', 'NOx', 'NH3', 'CO', 'SO2', 'O3', 'Benzene', 'Toluene', 'Xylene', 'AQI']
city_yearly_avg = city_data_agg.groupby(['State_Name', 'Year'])[pollutant_cols].mean().reset_index()

display(city_yearly_avg.head())
print(f"Shape of transformed city data: {city_yearly_avg.shape}")

crop_data_renamed = crop_data.rename(columns={'Crop_Year': 'Year'}).copy()
crop_data_renamed['State_Name'] = crop_data_renamed['State_Name'].str.lower().str.strip()
crop_yearly_sum = crop_data_renamed.groupby(['State_Name', 'Year'])[['Production', 'Area']].sum().reset_index()

display(crop_yearly_sum.head())
print(f"Shape of transformed crop data: {crop_yearly_sum.shape}")

merged_data = pd.merge(city_yearly_avg, crop_yearly_sum, on=['State_Name', 'Year'], how='outer')

display(merged_data.head())
print(f"Shape of merged data: {merged_data.shape}")

plt.figure(figsize=(16, 12))
sns.heatmap(merged_data.drop(columns=['Year']).corr(numeric_only=True), annot=True, cmap='coolwarm', fmt=".2f")
plt.title('Correlation Matrix of Air Quality and Crop Production Features')
plt.xticks(rotation=45, ha='right')
plt.yticks(rotation=0)
plt.tight_layout()
plt.show()
```

To merge city data treate air quality and crop data crop production both datasets are transformed to a State Name and Year

Cities are mapped to states   Year  is extracted from Date and then aggregated by State Name and Year using the mean for AQI and pollutant columns
crop data Crop Year is renamed to Year then aggregated by State Name and Year using the sum for Production and Area

An outer merge is performed on the aggregated DataFrames using State Name and Year

Higher AQI  correlate with lower crop Production and Area

Many pollutants  show strong positive correlations

```python
state_aqi = city_yearly_avg.groupby('State_Name')['AQI'].mean()
state_prod = crop_yearly_sum.groupby('State_Name')['Production'].sum()

most_polluted_state = state_aqi.idxmax()
least_polluted_state = state_aqi.idxmin()

print(f"Most polluted state: {most_polluted_state} (AQI: {state_aqi.max():.1f})")
print(f"Least polluted state: {least_polluted_state} (AQI: {state_aqi.min():.1f})")

prod_most = state_prod.get(most_polluted_state, 0)
prod_least = state_prod.get(least_polluted_state, 0)

fig, ax = plt.subplots(1, 2, figsize=(12, 5))

states = [most_polluted_state, least_polluted_state]
aqi_vals = [state_aqi[most_polluted_state], state_aqi[least_polluted_state]]
prod_vals = [prod_most, prod_least]
colors = ['red', 'green']

ax[0].bar(states, aqi_vals, color=colors, alpha=0.7)
ax[0].set_ylabel('Mean AQI')
ax[0].set_title('AQI: Most vs Least Polluted States')
ax[0].tick_params(axis='x', rotation=15)

ax[1].bar(states, prod_vals, color=colors, alpha=0.7)
ax[1].set_ylabel('Total Production')
ax[1].set_title('Crop Production: Most vs Least Polluted States')
ax[1].tick_params(axis='x', rotation=15)

plt.tight_layout()
plt.show()
```

The most polluted state Delhi produces less than the least polluted Meghalaya higher pollution correlates with specific seasons not uniform state-level industrialization

**TASK B**

```python
from scipy.stats import pearsonr

state_aqi_dict = city_yearly_avg.groupby('State_Name')['AQI'].mean().to_dict()
state_prod_dict = crop_yearly_sum.groupby('State_Name')['Production'].sum().to_dict()

aqi_states = set(state_aqi_dict.keys())
crop_states = set(state_prod_dict.keys())
common_states = sorted(list(aqi_states.intersection(crop_states)))

print(f"AQI states: {len(aqi_states)}")
print(f"Crop states: {len(crop_states)}")
print(f"Common states: {len(common_states)}")

state_aqi_vals = [state_aqi_dict[s] for s in common_states]
state_prod_vals = [state_prod_dict[s] for s in common_states]

correlation, p_value = pearsonr(state_aqi_vals, state_prod_vals)
print(f"\nPearson correlation (Common states AQI vs Production): {correlation:.3f}")
print(f"P-value: {p_value:.4f}")
print(f"R-squared (variance explained): {correlation**2:.3f}")

strength = "very weak" if abs(correlation) < 0.2 else ("weak" if abs(correlation) < 0.4 else ("moderate" if abs(correlation) < 0.7 else "strong"))
direction = "negative" if correlation < 0 else "positive"
print(f"\nInterpretation: {strength.capitalize()} {direction} correlation")

plt.figure(figsize=(10, 6))
plt.scatter(state_aqi_vals, state_prod_vals, alpha=0.6, s=100)

for i, state in enumerate(common_states):
    if i % 2 == 0:
        plt.annotate(state, (state_aqi_vals[i], state_prod_vals[i]), fontsize=8, alpha=0.5)

plt.xlabel('Mean AQI by State')
plt.ylabel('Total Production by State')
plt.title(f'State-level: AQI vs Crop Production (r={correlation:.3f}, p={p_value:.3f})')

z = np.polyfit(state_aqi_vals, state_prod_vals, 1)
p_func = np.poly1d(z)
sorted_aqi = np.sort(state_aqi_vals)
plt.plot(sorted_aqi, p_func(sorted_aqi), "r--", alpha=0.8, linewidth=2, label=f'Trend line')

plt.grid(alpha=0.3)
plt.legend()
plt.tight_layout()
plt.show()
```

AQI and production are NOT strongly linked at state level

The relationship explains only 4.6% of variation

Other factors rainfall, soil, irrigation farmer income technology likely have much larger effects
